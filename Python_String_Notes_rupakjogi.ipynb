## __Python "String Handbook"__

A **String** is a `basic, fundamental, or primitive data type` in Python. A string in Python is defined as any series of characters contained in *quotes {‘…’, “…”}*. Strings store entire sentences word by word.

### __Types of Quotes:__
1.   Single Quotes: '...'
2.   Double Quotes: "..."

```python
# @title Types of Quotes => 1) Single Quotes: '...'

s = 'Hello India'
print(s, '\nDatatype of s varible is', type(s))
```

```python
s1 = 'My name is Rupak C. Jogi'
print(s1, '\nDatatype of s1 varible is', type(s1))
```

```python
# @title Types of Quotes => 2) Double Quotes: "..."

d = "Happy Morning"
print(d, '\nDatatype of d varible is', type(d))
```

```python
d1 = "I am a Data Scientist"
print(d1, '\nDatatype of d1 varible is', type(d1))
```

### __Multi-line Strings:__


*   Single Triple Quotes: '''...'''
*   Double Triple Quotes: """..."""

>Triple quotes are used for multi-line strings or doc-strings.

```python
# @title Multi-line Strings: 1) Single Triple Quotes: '''...'''

ms = '''Python is an amazing programming language.
    It is widely used in web development, data science, and more.'''

print(ms, '\n\nDatatype of ms varible is', type(ms))
```

2.	**Slicing**
 - Slicing is used to `fetch a sub-string from a string`.
 1.	Positive Slicing
       - Uses *positive* indexing.
***Syntax:***

    string[starting_index : ending_index (excluded) : step_size]

*Where:*
- starting_index = The index from where we want to start slicing `(by default, it is 0)`.
- ending_index = The index till which we want the slice to be executed *(exclusive - n+1)*. `By default, it goes till the length of the string`.
- step_size = The number of jumps to make while slicing `(by default, the value of step_size is 1)`.


```python
a = 'Rupak Jogi'
print('Length of a is',len(a))
a[::]
```

```python
print(a[:6])
print(a[0:6])
print(a[0:6:1])
```

```python
print(a[6:])
print(a[6:len(a)+1])
print(a[12:len(a):1])
```

2.	Negative Slicing
***Syntax:***
     
     variable_name[starting_index : ending_index (excluded --> n+1) : step_size]

*Explanation:*
- In negative slicing, the *step_size is typically* -1. If not provided, `it defaults to 1.`
- The *starting_index* should be a negative index. If not provided, `it defaults to -1.`

> __Note:__
  When using slicing, the interpreter checks the step_size:
   - If **step_size** is 1 or *positive*, the `default starting_index is 0` and the `default ending_index is len(string).`
   - If **step_size** is -1 or *negative*, the `default starting_index is -1` and the `default ending_index is - (len(string) + 1).`
   - If the user has *not provided* a **step_size**, it `defaults to 1`, the `starting_index will be 0`, and the `ending_index will be len(string).`

> **Methods vs Functions**
- `Methods:` Methods are functions that are defined within a particular class (e.g., strings, lists, tuples, dictionaries, sets, frozensets).
- `Functions:` Functions are pieces of code that, once defined, can be called independently and an infinite number of times.

> **Method Characteristics:**
- Methods are associated with a class and are written inside the class.
- Methods cannot be called independently; they need an instance of the class to be invoked.

__casefold() ***Vs*** lower()__

- __lower():__ Converts **only ASCII** characters to `lowercase`.
- __casefold():__ Converts **both ASCII** *and* **non-ASCII** characters to `lowercase`.

```python
# Handling German Characters
# Using lower()
str3 = "Groß"
lower_str3 = str3.lower()
print('Using lower():', lower_str3)

# Using casefold()
str4 = "Groß"
casefold_str4 = str4.casefold()
print('Using casefold():', casefold_str4)
```

26.	isidentifier()
 - Returns ***True*** if the string is a valid identifier, otherwise returns ***False***.

>**Identifier**
  - An identifier is a name given to a variable, function, class, file, etc.

>__Properties of a *Valid Identifier*:__
1.	May contain alphanumeric characters `(A-Z, a-z, or 0-9)` and `underscores` (_).
2.	Must begin with an `alphabet` or an `underscore` (_) but never with a number.
3.	May contain numbers, but only after the `alphabet` or `underscore` (_).
4.	No spaces or special characters are allowed.
5.	Case-sensitive (e.g., a and A are treated differently).
